```python
import altair as alt

# Create a histogram showing the distribution of regular season win percentage
win_pct_distribution = alt.Chart(predict_data).mark_bar().encode(
    x=alt.X("reg_wins_pct")
        .bin(maxbins=20)
        .title("Regular-season Win Percentage (%)"),
    y=alt.Y("count()")
        .title("Number of Teams")
).properties(
    title="Distribution of Regular-season Win Percentage"
)

win_pct_distribution
```

```python
# Calculate the relative proportions of the 15 most frequent schools
school_proportions = (
    predict_data["school"]
    .value_counts(normalize=True)
    .head(15)
    .reset_index()
)
school_proportions.columns = ["school", "proportion"]

school_distribution = alt.Chart(school_proportions).mark_bar().encode(
    x=alt.X("school:N")
        .title("School"),
    y=alt.Y("proportion:Q")
        .title("Proportion of All Observations")
).properties(
    title="Proportion of Observations for the 15 Most Frequent Schools"
)

school_distribution
```

Visualization 1: This scatter plot shows the relationship between a team's regular season win percentage and its number of tournament wins. It is relevant to the research question because it allows us to examine whether teams with higher regular season win percentages tend to win more tournament games. The overall pattern can help us assess whether regular season performance may be useful for predicting tournament success.

```python
# Create a scatter plot showing the relationship between regular-season win percentage and tournament wins
scatter_plot = alt.Chart(predict_data).mark_circle().encode(
    x=alt.X("reg_wins_pct")
        .title("Regular-season Win Percentage (%)"),
    y=alt.Y("tourney_wins")
        .title("Number of Tournament Wins")
).properties(
    title="Regular-season Win Percentage vs. Tournament Wins"
)
scatter_plot
```

Visualization 2: This box plot compares the distribution of tournament wins across different regular season win percentage groups. The plot suggests that teams with higher regular-season win percentages tend to achieve more tournament wins. However, there is still variation in tournament performance within each group, suggesting that regular season win percentage may not fully explain tournament success.

```python
# Create a box plot showing tournament wins across regular-season win percentage groups
box_plot = alt.Chart(predict_data).mark_boxplot().encode(
    x=alt.X("reg_wins_pct")
        .bin(maxbins=8 )
        .title("Regular-season Win Percentage (%)"),
    y=alt.Y("tourney_wins")
        .title("Number of Tournament Wins")
).properties(
    title="Tournament Wins Across Regular-season Win Percentage Groups"
)

box_plot
```
